```python
operations = []

def ajouter_operation(date, libelle, compte_debit, compte_credit, montant):
    if montant <= 0:
        return "Erreur: montant invalide"

    operations.append({
        "Date": date,
        "Libelle": libelle,
        "Debit": compte_debit,
        "Credit": compte_credit,
        "Montant": montant
    })

    return "Operation ajoutee avec succes"

def afficher_journal():
    for op in operations:
        print(op)

def verifier_equilibre():
    total_debit = sum(op["Montant"] for op in operations)
    total_credit = sum(op["Montant"] for op in operations)

    return total_debit == total_credit

print("Bienvenue dans COMPTALIA")
print("Systeme comptable pret")
```

```python

if "operations" not in globals():
    operations = []

# 1. Ajouter une opération comptable
def ajouter_operation():
    print("=== AJOUTER UNE OPERATION ===")
    date = input("Date (AAAA-MM-JJ) : ")
    libelle = input("Libelle : ")
    debit = input("Compte debit : ")
    credit = input("Compte credit : ")
    montant = float(input("Montant : "))

    operations.append({
        "date": date,
        "libelle": libelle,
        "debit": debit,
        "credit": credit,
        "montant": montant
    })

    print("Operation ajoutee avec succes !")


# 2. Afficher le journal comptable
def afficher_journal():
    print("=== JOURNAL COMPTABLE ===")

    if not operations:
        print("Aucune operation.")
        return

    for op in operations:
        print("-----------------------")
        print("Date :", op["date"])
        print("Libelle :", op["libelle"])
        print("Debit :", op["debit"])
        print("Credit :", op["credit"])
        print("Montant :", op["montant"])


# 3. Calculer les soldes et les totaux
def calculer_soldes():
    soldes = {}

    for op in operations:
        debit = op["debit"]
        credit = op["credit"]
        montant = op["montant"]

        soldes[debit] = soldes.get(debit, 0) + montant
        soldes[credit] = soldes.get(credit, 0) - montant

    print("=== SOLDES DES COMPTES ===")
    for compte, solde in soldes.items():
        print(compte, ":", solde)

    print("Total des operations :",
          sum(op["montant"] for op in operations))


# 4. Detecter les erreurs comptables
def detecter_erreurs():
    print("=== CONTROLE DES ERREURS ===")
    erreurs = 0

    for op in operations:
        if op["montant"] <= 0:
            print("Erreur : montant invalide")
            erreurs += 1

        if not op["debit"] or not op["credit"]:
            print("Erreur : compte manquant")
            erreurs += 1

        if op["debit"] == op["credit"]:
            print("Erreur : debit et credit identiques")
            erreurs += 1

        if not op["date"] or not op["libelle"]:
            print("Erreur : date ou libelle manquant")
            erreurs += 1

    if erreurs == 0:
        print("Aucune erreur detectee !")
    else:
        print("Nombre d'erreurs :", erreurs)


print("Les 4 fonctions COMPTALIA sont pretes !")
```

```python

if "journal" not in globals():
    journal = []

def ajouter_operation():
    date = input("Date : ")
    libelle = input("Libelle : ")
    debit = float(input("Debit : "))
    credit = float(input("Credit : "))

    operation = {
        "date": date,
        "libelle": libelle,
        "debit": debit,
        "credit": credit
    }

    journal.append(operation)
    print("Operation ajoutee avec succes !")


def afficher_journal():
    print("\n===== JOURNAL COMPTABLE =====")
    for operation in journal:
        print(operation)
    if not journal:
        print("Journal vide.")


def calculer_solde():
    debit = sum(op["debit"] for op in journal)
    credit = sum(op["credit"] for op in journal)
    print("Total Debit :", debit)
    print("Total Credit :", credit)
    print("Solde :", debit - credit)


def verifier_equilibre():
    debit = sum(op["debit"] for op in journal)
    credit = sum(op["credit"] for op in journal)

    if debit == credit:
        print("Comptabilite equilibree !")
    else:
        print("Erreur : Debit different du Credit !")


def rechercher_operation():
    nom = input("Libelle a rechercher : ").lower()

    for op in journal:
        if op["libelle"].lower() == nom:
            print("Operation trouvee :", op)
            return

    print("Operation introuvable.")


def supprimer_operation():
    nom = input("Libelle a supprimer : ").lower()

    for op in journal:
        if op["libelle"].lower() == nom:
            journal.remove(op)
            print("Operation supprimee avec succes !")
            return

    print("Operation introuvable.")


def comptalia():
    while True:
        print("\n========== COMPTALIA ==========")
        print("1. Ajouter une operation")
        print("2. Afficher le journal")
        print("3. Calculer le solde")
        print("4. Verifier l'equilibre")
        print("5. Rechercher une operation")
        print("6. Supprimer une operation")
        print("0. Quitter")
        print("==============================")

        choix = input("Votre choix : ")

        if choix == "1":
            ajouter_operation()
        elif choix == "2":
            afficher_journal()
        elif choix == "3":
            calculer_solde()
        elif choix == "4":
            verifier_equilibre()
        elif choix == "5":
            rechercher_operation()
        elif choix == "6":
            supprimer_operation()
        elif choix == "0":
            print("Merci d'utiliser COMPTALIA !")
            break
        else:
            print("Choix invalide !")


comptalia()
```
